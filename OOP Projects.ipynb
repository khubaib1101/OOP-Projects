```python
# Task 1:
class Bank:
  def get_data(self,name,acc_no):
    self.name= name
    self.acc_no= acc_no

    if acc_no == 1234:
      self.initial_balance = 50000
    elif acc_no == 5678:
      self.initial_balance = 60000
    else:
      self.initial_balance = 0
  def info(self):
    print("Name",self.name)
    print("Account Number",self.acc_no)
    print("balance",self.initial_balance)
name = input("Enter your Name: ")
acc_no = int(input("Enter your Account Number: "))
a= Bank()
a.get_data(name,acc_no)
a.info()
```

```python
# Task 2:
class Student:
  def get_data(self,name,roll_no):
    self.name = name
    self.roll_no = roll_no
    if roll_no == 1:
      self.course = "Computer Science"
    elif roll_no == 2:
      self.course = "Information Technology"
    else:
      self.course = "None"

  def info(self):
    print("Name",self.name)
    print("ROll No",self.roll_no)
    print("Course",self.course)
name = input("Enter your Name")
roll_no = int(input("Enter your roll number"))

s = Student()
s.get_data(name,roll_no)
s.info()
```
